```python
import altair as alt 
import numpy as np 
import pandas as pd 

df = pd.DataFrame({
    "Concentration": [0.10, 0.15, 0.20, 0.30, 0.50, 0.70, 0.95],
    "Number of Insects": [47, 53, 55, 52, 46, 54, 52],
    "Number Killed": [8, 14, 24, 32, 38, 50, 50],
    "Percent Killed": [17.0, 26.4, 43.6, 61.5, 82.6, 92.6, 96.2]
})

df.head()
```

```python
alt.Chart(df).mark_point().encode(
    x = alt.X("Concentration"),
    y = alt.Y("Number Killed")
)
```
